## ANP Analytics - Perguntas de Negócios

```sql
SELECT * FROM default.gold_combustivel LIMIT 5;
```

### 1 - Quais estados apresentam o maior preço médio dos combustíveis em 2025?

Objetivo de negócio:

Identificar os estados onde o consumidor paga mais caro pelos combustíveis.

```sql
SELECT 
    Estado_Sigla,
    ROUND(AVG(Valor_Venda), 2) AS preco_medio
FROM default.gold_combustivel
GROUP BY Estado_Sigla
ORDER BY preco_medio DESC
LIMIT 5;
```

## 2 - Quais estados apresentam o menor preço médio dos combustíveis em 2025?

```sql
SELECT 
    Estado_Sigla,
    ROUND(AVG(Valor_Venda), 2) AS preco_medio
FROM default.gold_combustivel
GROUP BY Estado_Sigla
ORDER BY preco_medio
LIMIT 5;
```

### Pergunta 3 - Como os preços evoluíram ao longo do tempo?

```sql
SELECT 
    Ano_Mes,
    ROUND(AVG(Valor_Venda), 2) AS preco_medio
FROM default.gold_combustivel
GROUP BY Ano_Mes
ORDER BY Ano_Mes
```

## 4 - Qual combustível possui o maior preço médio?

```sql
SELECT
    Produto,  
    ROUND(AVG(Valor_Venda), 2) AS preco_medio_produto
FROM default.gold_combustivel
GROUP BY Produto
ORDER BY preco_medio_produto DESC;
```

### 5 - Em quais estados a diferença entre Gasolina e Etanol é maior?

```sql
SELECT 
    Estado_Sigla,
    preco_medio_gasolina, 
    preco_medio_etanol,
    preco_medio_gasolina - preco_medio_etanol AS direfenca_preco
FROM (
    SELECT 
        Estado_Sigla,
    ROUND(
        AVG(
            CASE
                WHEN Produto = 'GASOLINA' THEN Valor_Venda END), 2) AS preco_medio_gasolina,
    ROUND(
        AVG(
            CASE
                WHEN Produto = 'ETANOL' THEN Valor_Venda END), 2) AS preco_medio_etanol
    FROM default.gold_combustivel
    WHERE Produto IN ('GASOLINA', 'ETANOL')
    GROUP BY Estado_Sigla
    ) t
ORDER BY direfenca_preco DESC
LIMIT 10;
```

### 6 Quais bandeiras apresentam os maiores preços médios dos combustíveis em 2025?

```sql
SELECT 
    Bandeira,
    ROUND(AVG(Valor_Venda), 2) AS preco_medio
FROM default.gold_combustivel
GROUP BY Bandeira
ORDER BY preco_medio DESC
LIMIT 5;
```
